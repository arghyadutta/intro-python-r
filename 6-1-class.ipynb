# Class
- A template for creating objects that bundles data (attributes) and behavior (methods) together. 
- define as `class ClassName:`
- the `__init__` method sets up the initial attributes when an object is created
- Each object made from the class is called an instance
- `self` refers to that specific instance inside the class's methods, letting each object keep its own data while sharing the same behavior.

```python
s = 'string'
s.capitalize()
a = [1,3,4]
print(type(a))
```

```python
class Person:
    def __init__(self, a):
        self.name = a

    def find_name_length(self):
        print(f'The length of {self.name} is {len(self.name)}')

p1 = Person('debjit')
print(p1.name)
p1.find_name_length()
```

```python
class Customer:
    def __init__(self, name, age, current_balance=0):
        self.name = name
        self.age = age
        self.current_balance = current_balance

    def change_balance(self, x):
        self.current_balance = self.current_balance + x
```

```python
person1 = Customer("Alice", 25)

# person1.__dir__()
person1.name
person1.age
person1.current_balance
deposit = -10 # real banks will do something more decent here :D 
person1.change_balance(deposit)
person1.current_balance
```

```python
person2 = Customer("Bob", 20, 56)
# type(person2)
person2.name
person2.age
person2.current_balance
deposit = 10 # real banks will do something more decent here :D 
person2.change_balance(deposit)
person2.current_balance
```

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def is_passing(self):
        return self.grade >= 50
```

```python
student = Student("John", 5)
print(student.name)
print(student.grade)
print(student.is_passing())
```
