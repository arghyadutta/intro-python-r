*Why use Pandas?*

While NumPy is excellent for numerical matrices (homogeneous data), Pandas is designed for tabular data (heterogeneous data). 
Think of a DataFrame as a NumPy array with labels, or an Excel sheet, allowing you to mix strings, integers, and floats in one table.

```python
import numpy as np
import pandas as pd
```

## Pandas Series

A 1D labeled array that can hold any data type (ints, floats, strings, objects).

- Like a NumPy 1D array, but with an index
- Each value is accessed by its label, not just position
- Supports vectorized operations and alignment by index

```python
foo = [10, 20, 30]
foo[0]
```

```python
# s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])

s = pd.Series({'a':10,'b':20,'c':30})

# the values of a series are returned as numpy arrays
print(s, s.values, type(s.values))

# vectorized operation (often intuitive)
s + 5
```

## DataFrame

2D, labeled tabular data structure.

- Rows and columns both have indexes
- Columns can hold different data types
- Supports vectorized operations, alignment, grouping, merging

![](Pandas_dataframe.png)

[Source: Wikimedia commons](https://commons.wikimedia.org/wiki/File:Pandas_dataframe.png)

```python
# defining a dataframe from a dictionary
data = {
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'score': [88.5, 92.0, 79.5]
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)

# print("the structure of your data")
# df.info()

print("get some statistical values")
print(df.describe())

print(f"\nColumns: {df.columns.tolist()}")
```

## Indexing: loc vs iloc

- `.loc`: Label-based (uses names/labels)
- `.iloc`: Integer-position based (uses 0, 1, 2...)

```python
# row
df
df.loc[0]['name'] # label-based
df.iloc[0] # position-based

# column
df.iloc[:,1]
df["age"]
df[["name", "score"]]
```

```python
# boolean indexing works as in numpy
df
# df[df['age'] > 28]
df[df['score'] > 90]
df.query("score >= 90")

# more complex logical operation

print(df[(df['age'] > 25) & (df['score'] > 80)])
```

## Operations and column manipulations

```python
# New column
df
df["passed"] = df["score"] >= 85
df["old"] = df["age"] > 25
df

# Vectorized (preferred)
df["age_squared"] = df["age"] ** 2

df.sort_values("age", ascending=False)
```

```python

df.head(2) #first two rows
df.tail(2) # last two rows

# how many unique values in a particular column?
df.value_counts("passed")
```

## GroupBy

```python
df['department'] = ['CS', 'Math', 'CS']
df
df.groupby('department')['score'].mean()
```

## Merge

```python
info = pd.DataFrame({
    'name': ['Charlie','Alice', 'Bob'],
    'city': ['Chennai','Delhi', 'Mumbai']
})
df
info
pd.merge(df, info, on='name')
```
